```python
import csv
import random

def get_student(filename, sample_size):
    students = []
    
    with open(filename) as file:
        reader = csv.reader(file)
        for line in reader:
            if line:
                students.append(line[0])
        
    return random.sample(students, k=sample_size)

print(get_student("studenti.csv",4))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sim(p=0.5):
    N = 0
    S = 0
    while True:
        N += 1
        stema = random.random() < p
        if stema:
            z = random.randint(1, 6)
            S += (z - 3)
            break
        else:
            S -= 0.5
    return N, S

def monte_carlo(sample_size, p=0.5):
    res = []
    
    for _ in range(sample_size):
        _, S = sim(p)
        res.append(S)
        
    mean_s = np.mean(res)
    
    plt.hist(res, edgecolor='black')
    plt.show()

    print(mean_s)
    
    return mean_s

N, S = sim()

c = monte_carlo(10000, 0.5)

d1 = monte_carlo(10000, 0.3)
d2 = monte_carlo(10000, 0.7)
```

```python
import seaborn as sns

lambdas = np.array([3, 6, 4])
prob = [3/13, 6/13, 4/13]
N = 10000

frizeri_alesi = np.random.choice([0, 1, 2], size=N, p=prob)
scales = 1 / lambdas[frizeri_alesi]

X = np.random.exponential(scale=scales)

mean_X = np.mean(X)
deviatia_std_X = np.std(X)

print(f"Medie: {mean_X}")
print(f"Deviatia std: {deviatia_std_X}")

sns.kdeplot(X)
plt.show()
```
