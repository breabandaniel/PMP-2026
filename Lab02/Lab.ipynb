```python
PB = 0.01
Sens = 0.95
Spec = 0.9

PPoz_NB = 1 - Spec
PNB = 1 - PB

PPoz = PPoz_NB * PNB + Sens * PB

PB_Poz = Sens * PB / PPoz

print(PB_Poz)
```

```python
import numpy as np
import random

def sim(sample_size=1000000):
    bolnavi = np.random.binomial(n=sample_size, p=0.01)
    sanatosi = sample_size - bolnavi

    true_poz = 0
    total_poz = 0
    
    for _ in range(bolnavi):
        if random.random() < 0.95:
            true_poz += 1
            total_poz += 1
            
    for _ in range(sanatosi):
        if random.random() < 0.10:
            total_poz += 1

    prob = true_poz / total_poz

    print(prob)

sim()
```
