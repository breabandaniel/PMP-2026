```python
import random

def f():
    urna = ['R', 'R', 'R', 'A', 'A', 'A', 'A', 'N', 'N']

    zar = random.randint(1, 6)
    
    if zar in [2, 3, 5]:
        urna.append('N')
    elif zar == 6:
        urna.append('R')
    else: 
        urna.append('A')
        
    return random.choice(urna)
```

```python
simulari = 100000
reds = 0

for _ in range(simulari):
    res = f()
    if res == 'R':
        reds += 1

prob_estimata = reds / simulari
print(prob_estimata)
```
