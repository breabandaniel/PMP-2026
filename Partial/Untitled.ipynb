```python
#1a

import random
import numpy as np
from hmmlearn.hmm import CategoricalHMM

start_prob = np.array([1/4, 1/4, 1/4, 1/4])

trans_mat = np.array([
    [0.2, 0.4, 0, 0.4],  #1 
    [0.4, 0.2, 0.4, 0],  #2 
    [0, 0.4, 0.2, 0.4],  #3 
    [0.4, 0, 0.4, 0.2]  #4 
])

emit_mat = np.array([
    [0.7, 0.2, 0.1],  #1 
    [0.2, 0.6, 0.2],  #2 
    [0.1, 0.3, 0.6],  #3 
    [0.5, 0.2, 0.3],  #4
])

model = CategoricalHMM(n_components=4)
model.startprob_ = start_prob
model.transmat_ = trans_mat
model.emissionprob_ = emit_mat

target_seq = [0, 1, 2, 2, 0]
obs_seq = np.array(target_seq).reshape(-1, 1)

log_prob_viterbi, hidden_states = model.decode(obs_seq, algorithm="viterbi")

print(f"{hidden_states=} (rezultatul o sa fie de forma camera-1)")
print(f"Probabilitatea = {np.exp(log_prob_viterbi)}")
```

```python
#3a

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

theta = np.linspace(0, 1, 500)

prior = beta.pdf(theta, 2, 8)
post = beta.pdf(theta, 7, 23)

plt.figure(figsize=(8, 5))
plt.plot(theta, prior, label="prior: Beta(2, 8)")
plt.plot(theta, post, label="posterior: Beta(7, 23)", linewidth=2.5)

plt.legend()
plt.show()
```
