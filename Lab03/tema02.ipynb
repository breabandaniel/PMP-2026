```python
import pandas as pd

data = {
    'Program': ['Prog 1', 'Prog 1', 'Prog 2', 'Prog 2'],
    'Disciplina': ['A', 'B', 'A', 'B'],
    'Promovati': [80, 9, 1, 20],
    'Evaluati': [100, 10, 10, 100]
}

df = pd.DataFrame(data)
df['Rata'] = (df['Promovati'] / df['Evaluati']) * 100

print(df[['Program', 'Disciplina', 'Rata']].to_string())

print("\nRATA GLOBALA")
df_global = df.groupby('Disciplina')[['Promovati', 'Evaluati']].sum().reset_index()
df_global['Rata_Globala'] = (df_global['Promovati'] / df_global['Evaluati']) * 100
print(df_global[['Disciplina', 'Rata_Globala']].to_string())
```
