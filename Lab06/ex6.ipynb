```python
import numpy as np
import arviz as az
import matplotlib.pyplot as plt
from scipy.stats import gamma

k = 18
n = 1

alpha_prior = 0.18
beta_prior = 0.01

alpha_post = alpha_prior + k
beta_post = beta_prior + n

modul = (alpha_post - 1) / beta_post
print(f"{modul=}")

a = alpha_post
scale = 1 / beta_post

x = np.linspace(gamma.ppf(0.001, a, scale=scale), gamma.ppf(0.999, a, scale=scale), 500)
y = gamma.pdf(x, a, scale=scale)

plt.plot(x, y)
limita_jos, limita_sus = gamma.interval(0.94, a, scale=scale)
plt.fill_between(x, y, where=((x >= limita_jos) & (x <= limita_sus)))
plt.show()
```
